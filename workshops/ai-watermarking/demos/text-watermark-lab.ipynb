## 2. Give our model four words
`import random` gives us Python's tool for picking things at random.

Square brackets make a **list**. Here, the list contains four words.

```python
import random

words = ["calm", "quiet", "bright", "sunny"]
print(words)
```

`random.choice()` picks one item from a list. Run this cell a few times. The choice may change.

```python
print(random.choice(words))
```

## 3. Pick 40 words
We'll repeat the same action 40 times. A `for` loop does the repeating.

- `[]` starts an empty list.
- `append()` adds one choice to that list.
- The colon starts the loop's instructions. The indented line runs each time around it.
- `" ".join()` puts spaces between the chosen words.

The seed `7` makes this example repeatable. Leave it alone for now.

```python
random.seed(7)
normal_words = []
for turn in range(40):
    normal_words.append(random.choice(words))
normal_text = " ".join(normal_words)
print(normal_text)
```

## 4. Add a preference
We want **calm** and **bright** to appear more often.

Imagine picking a ticket from a bag. Put three tickets for each preferred word in the bag, and one for each other word. All four words can still appear.

This preference is our simple watermark rule.

```python
biased_bag = ["calm", "calm", "calm", "quiet",
              "bright", "bright", "bright", "sunny"]
```

Run the same loop using the new bag. The only change to the choice itself is `words` becoming `biased_bag`.

```python
random.seed(7)
marked_words = []
for turn in range(40):
    marked_words.append(random.choice(biased_bag))
marked_text = " ".join(marked_words)
print(marked_text)
```

## 5. Look for the preference
`count()` tells us how often a word appears. Add the counts for our two preferred words.

Before running: which list should have more of them?

```python
normal_count = normal_words.count("calm") + normal_words.count("bright")
marked_count = marked_words.count("calm") + marked_words.count("bright")
print("Normal:", normal_count, "out of 40")
print("Marked:", marked_count, "out of 40")
```

With these cells unchanged, you should see **23** for normal and **33** for marked.

The normal text contains preferred words too. We're looking for an unusually frequent pattern across many choices. A count on its own does not prove who wrote something.

## 6. Save the text, then read it back
`Path` gives us a way to work with a file. `write_text()` saves our words. `read_text()` reads them back.

This creates `marked-words.txt` in the notebook's working folder.

```python
from pathlib import Path

text_file = Path("marked-words.txt")
text_file.write_text(marked_text, encoding="utf-8")
received_text = text_file.read_text(encoding="utf-8")
print(received_text)
```

`split()` separates the saved text into words again. Check that the preference is still there.

```python
received_words = received_text.split()
received_count = received_words.count("calm") + received_words.count("bright")
print("From the file:", received_count, "out of 40")
```

## 7. Change the words
`replace()` swaps one piece of text for another. We'll replace every preferred word with one of the other words.

Predict what the count will become, then run this.

```python
edited_text = received_text.replace("calm", "quiet")
edited_text = edited_text.replace("bright", "sunny")
edited_words = edited_text.split()
edited_count = edited_words.count("calm") + edited_words.count("bright")
print(edited_text)
print("After editing:", edited_count, "out of 40")
```

The count is now **0**. This simple rule was easy to remove because we knew exactly which words to replace.
