## 1. Make a picture
`import` brings in a tool we can use. `Image` works with pictures. `display()` shows them in the notebook.

```python
from PIL import Image
from IPython.display import display
```

A picture contains little squares called **pixels**. This one is 320 pixels wide and 180 pixels high.

`RGB` means red, green and blue. Each number goes from 0 to 255. Our picture uses the same colour everywhere: `(200, 210, 230)`.

```python
picture = Image.new("RGB", (320, 180), (200, 210, 230))
display(picture)
```

## 2. Look at one pixel
Positions start at zero. `(0, 0)` is the top-left pixel.

`getpixel()` reads its three colour numbers. You should see `(200, 210, 230)`.

```python
print(picture.getpixel((0, 0)))
```

Let's change its red value from **200** to **201**. `copy()` keeps our original safe. `putpixel()` changes a pixel in the copy.

The numbers change by one. Visually, the picture should look almost the same.

```python
marked_picture = picture.copy()
marked_picture.putpixel((0, 0), (201, 210, 230))
print("Original pixel:", picture.getpixel((0, 0)))
print("Changed pixel:", marked_picture.getpixel((0, 0)))
```

## 3. Choose a tiny message
Our message is a list of four digits: **1, 0, 1, 1**. A digit that can be only 0 or 1 is called a **bit**.

We'll put one bit into each of the first four pixels.

| Bit | Red value we'll save |
| --- | --- |
| 0 | 200 |
| 1 | 201 |

This works because our plain picture starts with a red value of 200 everywhere.

```python
message = [1, 0, 1, 1]
print("Our message:", message)
```

## 4. Put the message into the picture
A `for` loop repeats the indented lines. `range(4)` gives us positions 0, 1, 2 and 3.

`message[x]` reads the bit at position `x`. Adding it to 200 gives us either 200 or 201. The green and blue values stay the same.

```python
marked_picture = picture.copy()
for x in range(4):
    red = 200 + message[x]
    marked_picture.putpixel((x, 0), (red, 210, 230))
display(marked_picture)
```

## 5. Save it, then open the saved file
`save()` writes an image file. PNG preserves the pixel values we're using.

We'll save the original too, so we have something to compare with.

```python
picture.save("original.png")
marked_picture.save("marked.png")
print("Saved original.png and marked.png")
```

`Image.open()` reads a picture from a file. We will check this saved copy.

```python
received_picture = Image.open("marked.png")
display(received_picture)
```

## 6. Read our message back
`%` gives the remainder after division. For our red values:

- `200 % 2` is **0**, because 200 is even.
- `201 % 2` is **1**, because 201 is odd.

The names `red, green, blue` receive the three numbers from `getpixel()`. We'll keep the red value and add its remainder to a list. `append()` adds one item to a list. `==` checks whether two things are equal.

```python
recovered = []
for x in range(4):
    red, green, blue = received_picture.getpixel((x, 0))
    recovered.append(red % 2)
print("We hid:", message)
print("We read:", recovered)
print("Same message?", recovered == message)
```

You should see `[1, 0, 1, 1]` and `True`.

What would we read from the original picture? Try the same steps. Every red value there is 200.

```python
original_picture = Image.open("original.png")
original_bits = []
for x in range(4):
    red, green, blue = original_picture.getpixel((x, 0))
    original_bits.append(red % 2)
print("Original picture:", original_bits)
```

## 7. Try a JPEG
JPEG makes a smaller file by changing some image information. Those changes can include our hidden digits.

We'll use a quality setting of 20. Before running: will the picture still look similar? Will the message survive?

```python
received_picture.save("compressed.jpg", quality=20)
compressed_picture = Image.open("compressed.jpg")
display(compressed_picture)
```

Read the same four pixels again. The code is the same; only the picture we're reading has changed.

```python
jpeg_bits = []
for x in range(4):
    red, green, blue = compressed_picture.getpixel((x, 0))
    jpeg_bits.append(red % 2)
print("Before JPEG:", message)
print("After JPEG:", jpeg_bits)
print("Same message?", jpeg_bits == message)
```

With the starting picture and message unchanged, the JPEG no longer carries the same four bits. The picture can look fine while the hidden message changes.

```python
practice_message = [0, 1, 0, 1]
practice_picture = picture.copy()
for x in range(4):
    red = 200 + practice_message[x]
    practice_picture.putpixel((x, 0), (red, 210, 230))
practice_picture.save("practice.png")
```

Open that file and read its four bits. This repeats the same steps you just learned.

```python
practice_file = Image.open("practice.png")
practice_bits = []
for x in range(4):
    red, green, blue = practice_file.getpixel((x, 0))
    practice_bits.append(red % 2)
print("New message:", practice_bits)
print("Same message?", practice_bits == practice_message)
```

You should read `[0, 1, 0, 1]` and `True`.

## What we built
We stored four bits in pixel values, saved the picture, and read the bits back. JPEG showed how easily this particular method breaks.

Four bits can also appear by chance. This is a way to learn the mechanism, not evidence of who created a picture. The learned image watermarks discussed in the workshop use much more of the image.
